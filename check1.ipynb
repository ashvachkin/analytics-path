```python
patients = [
    {"name": "Иванов", "age": 54, "temp": 38.2, "dept": "терапия"},
    {"name": "Петрова", "age": 31, "temp": 36.6, "dept": "хирургия"},
    {"name": "Сидоров", "age": 67, "temp": 37.5, "dept": "терапия"},
    {"name": "Козлова", "age": 45, "temp": 39.1, "dept": "кардиология"},
    {"name": "Орлов", "age": 29, "temp": 36.9, "dept": "хирургия"},
]
```

```python

def fever (patients,threshold=37.5):
    result = []
    for patient in patients:
        if patient["temp"] >= threshold:
            result.append(patient["name"])
    return result
print(fever(patients))
```

```python
def avg_age(patients):
    count = 0 
    total = 0
    for patient in patients:
       total += patient["age"]
       count += 1 
    if count == 0:
        return None
    return total / count
print(avg_age([]))
print(avg_age(patients))

```

```python
def count_by_dept(patients):
    counts = {}
    for patient in patients:
        dept = patient["dept"]
        if dept in counts:
            counts[dept] += 1      
        else:
            counts[dept] = 1     
    return counts
print(count_by_dept(patients))

```
