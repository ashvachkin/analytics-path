## Часть 1
- my_mean([]) возвращает None, а не 0 потому что нули это реальные данные, а None показывает их отсутствие 
- my_sum([]) возвращает 0 потому что сумма пустого набора равна нулю


```python
raw = "132, 118.5, 145, 97, 160, 124, 138.2"
```

```python
def parse_numbers(text):
    result = [] 
    if text.strip() == "":
        return []                      
    for elem in text.split(","):
        result.append(float(elem))
    return result                     
numbers = parse_numbers(raw)          
print(numbers)
```

```python
def my_count(numbers):
    count = 0
    for elem in numbers:
        count += 1
    return count
print(my_count(numbers))
```

```python
def my_sum(numbers):
    sum_result = 0
    for elem in numbers:
        sum_result += elem
    return sum_result
print(my_sum(numbers))
```

```python
def my_mean(numbers):
    if not numbers:
        return None
    return my_sum(numbers) / my_count(numbers)
print(my_mean(numbers))
```

```python
def my_min(numbers):
    if not numbers:
        return None
    acc = numbers[0]
    for elem in numbers:
        if elem < acc:
            acc = elem
    return acc
print(my_min(numbers))
```

```python
def my_max(numbers):
    if not numbers :
        return None
    acc = numbers[0]
    for elem in numbers:
        if elem > acc:
            acc = elem
    return acc
print(my_max(numbers))
```

```python
def report(numbers):
    mean = my_mean(numbers)
    if mean is not None:
        mean = round(mean,2)
    print(f"Количество: {my_count(numbers)} \nСумма: {my_sum(numbers)} \nСреднее: {mean} \nМин: {my_min(numbers)} \nМакс: {my_max(numbers)}")
report(numbers)
```

```python
test_cases = [raw, "", "   ", "120", "-5, -2, -9"]

for case in test_cases:
    print(f"Набор: '{case}'")
    report(parse_numbers(case))
    print()
```

## Часть 2
- top_code() возвращает первый код который набрал больший счетчик, это проблема тк может быть 2 кода с одинаковыми    значениями и по хорошему надо выводить оба 

```python
visits = ["E11", "I10", "I10", "I10", "E11", "E11", "M54", "E11", "I10"]
```

```python
def count_codes(visits):
    result = {}
    for code in visits:
        if code not in result:
            result[code] = 1
        else:
            result[code] += 1 
    return result
print(count_codes(visits))
```

```python
def top_code(counts):
    if not counts:
        return None
    best_code = None
    best_count = 0
    for code, count in counts.items():      
        if count > best_count :
            best_code = code
            best_count = count
    return best_code 
print(top_code(count_codes(visits)))
print(count_codes([]))   
print(top_code({}))    

  
```
